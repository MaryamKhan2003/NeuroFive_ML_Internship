### 📊 Initial Data Exploration Summary

The Titanic dataset contains 891 rows and 12 columns, representing passenger details such as age, gender, ticket class, and survival status. Initial inspection shows that some columns, particularly **Age**, **Cabin**, and **Embarked**, contain missing values, with Cabin having the highest proportion of missing data.

The dataset includes a mix of numerical features (such as Age, Fare, and family-related counts) and categorical features (such as Sex and Embarked). The target variable **Survived** indicates whether a passenger lived or died.

Overall, the dataset is well-structured but requires preprocessing, especially handling missing values and encoding categorical variables, before applying machine learning models. 

```python

# ================================
# 1. Import Libraries
# ================================
import pandas as pd

# ================================
# 2. Load Dataset
# ================================
# Make sure train.csv is in the same folder
df = pd.read_csv("train.csv")


# ================================
# 3. Basic Inspection
# ================================

print("First 5 Rows:")
print(df.head())

print("\nDataset Info:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

# ================================
# 4. Shape of Dataset
# ================================
rows, cols = df.shape
print("\nDataset Shape:")
print("Rows:", rows)
print("Columns:", cols)

# ================================
# 5. Missing Values
# ================================
print("\nMissing Values in Each Column:")
print(df.isnull().sum())

# ================================
# 6. Column Types
# ================================

# Numerical Columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
print("\nNumerical Columns:")
print(numerical_cols)

# Categorical Columns
categorical_cols = df.select_dtypes(include=['object']).columns
print("\nCategorical Columns:")
print(categorical_cols)



# ================================
# 7. Survival Distribution
# ================================
print("\nSurvival Count:")
print(df['Survived'].value_counts())



# ================================
# 8. Summary Statements (Console)
# ================================
print("\nSummary Insights:")
print(f"The dataset contains {rows} rows and {cols} columns.")
print("Missing values are present in Age, Cabin, and Embarked columns.")
print("Numerical features include:", list(numerical_cols))
print("Categorical features include:", list(categorical_cols))
```
